```python
import pandas as pd
from evidently import Report
from evidently.metrics import *
from evidently.presets import *
```

```python
REF_CSV = "../../data/processed/train.csv"
CUR_CSV = "../../data/processed/test.csv"

reference_data = pd.read_csv(REF_CSV)
current_data = pd.read_csv(CUR_CSV)
```

```python
report = Report(metrics=[
        DataDriftPreset(),
        DataSummaryPreset()
    ])
```

```python
print("Analyzing data drift... this may take a moment.")
result = report.run(reference_data=reference_data, current_data=current_data)
```

```python
result.save_html("output_file.html")
print(f"Report generated successfully: output_file.html")
```
